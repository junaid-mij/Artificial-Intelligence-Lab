To learn the basic features of the NumPy library, including array creation, numerical operations, and matrix manipulation in Python.

---

## <font color="green"><b>Program Code</b></font>

```python
import numpy as np

# Create a NumPy array
numbers = np.array([10, 20, 30, 40, 50])

# Display array
print("Array:", numbers)

# Numerical operations
print("Maximum Value:", np.max(numbers))
print("Minimum Value:", np.min(numbers))
print("Average:", np.mean(numbers))

# Create a 2x3 matrix
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

print("Matrix:")
print(matrix)
```

---

## <font color="purple"><b>Code Explanation</b></font>

### <font color="blue">Import NumPy Library</font>

```python
import numpy as np
```

Imports the NumPy library and assigns it the alias `np`.

### <font color="blue">Array Creation</font>

```python
numbers = np.array([10, 20, 30, 40, 50])
```

Creates a NumPy array containing five numbers.

### <font color="blue">Display Array</font>

```python
print("Array:", numbers)
```

Displays all elements of the array.

### <font color="blue">Numerical Operations</font>

```python
print("Maximum Value:", np.max(numbers))
print("Minimum Value:", np.min(numbers))
print("Average:", np.mean(numbers))
```

- `np.max()` finds the largest value.
- `np.min()` finds the smallest value.
- `np.mean()` calculates the average value.

### <font color="blue">Matrix Creation</font>

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
```

Creates a two-dimensional NumPy matrix with 2 rows and 3 columns.

### <font color="blue">Display Matrix</font>

```python
print(matrix)
```

Displays the matrix elements.
