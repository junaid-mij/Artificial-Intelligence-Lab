### <font color="green">Objective:</font>

To perform numerical operations using the NumPy library.

# <font color="red">Program</font>

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Array:", numbers)
print("Maximum Value:", np.max(numbers))
print("Minimum Value:", np.min(numbers))
print("Average:", np.mean(numbers))

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matrix:")
print(matrix)
```
