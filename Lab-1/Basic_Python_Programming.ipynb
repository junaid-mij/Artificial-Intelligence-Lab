# ***Objective***

To learn the basic concepts of Python programming such as variables, mathematical operations, loops, and functions using Google Colab.

````python
# Basic Python Program

# Store two numbers in variables
num1 = 12
num2 = 8

# Calculate multiplication
product = num1 * num2
print("Product:", product)

# Loop to print numbers from 1 to 5
for number in range(1, 6):
    print("Count:", number)

# Define a function to find the cube of a number
def cube(n):
    return n ** 3

# Call the function
print("Cube of 3:", cube(3))
````

## <font color="teal">Code Explanation</font>

## Variable Declaration

```python
num1 = 12
num2 = 8
```

Two variables are created and assigned values.

## Multiplication Operation

```python
product = num1 * num2
print("Product:", product)
```

Multiplies the two numbers and displays the result.

## Loop

```python
for number in range(1, 6):
    print("Count:", number)
```

The loop prints numbers from 1 to 5.

## Function Definition

```python
def cube(n):
    return n ** 3
```

A function named `cube()` is created to calculate the cube of a number.

## Function Call

```python
print("Cube of 3:", cube(3))
```

Calls the function with the value 3 and displays the result.
