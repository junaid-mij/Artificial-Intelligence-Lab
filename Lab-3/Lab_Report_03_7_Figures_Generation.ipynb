# <font color="darkblue" size="6"><b>Graphs (With Explanation) - 7 Figures Generation</b></font>

### <font color="green">Using:</font>

- Matplotlib
- Seaborn

```python
import matplotlib.pyplot as plt   # Used for plotting graphs
import seaborn as sns             # Used for advanced visualization
import numpy as np                # Used for generating data
```

# <font color="red">1. Line Plot</font>

Displays data points connected by lines to show trends.

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

# <font color="red">2. Bar Chart</font>

Compares values among different categories.

```python
categories = ['A', 'B', 'C']
values = [5, 7, 3]

plt.bar(categories, values)
plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Values")
plt.show()
```

# <font color="red">3. Histogram</font>

Shows the frequency distribution of data.

```python
data = np.random.randn(100)

plt.hist(data)
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# <font color="red">4. Scatter Plot</font>

Shows the relationship between two variables.

```python
x = [1, 2, 3]
y = [4, 5, 6]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

# <font color="red">5. Box Plot</font>

Displays data distribution and outliers.

```python
data = [10, 20, 30, 40, 50]

sns.boxplot(x=data)
plt.title("Box Plot")
plt.show()
```

# <font color="red">6. Heatmap</font>

Uses colors to represent values in a matrix.

```python
data = np.random.rand(5,5)

sns.heatmap(data, annot=True)
plt.title("Heatmap")
plt.show()
```

# <font color="red">7. Pie Chart</font>

Represents data as portions of a circle.

```python
values = [10, 20, 30]
labels = ['A', 'B', 'C']

plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```
